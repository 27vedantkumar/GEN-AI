# Pandas Crash Course Cheat Sheet

### 25 Most Useful Pandas Functions

| Category            | Function / Method     | Purpose                         | Example                                                |
| ------------------- | --------------------- | ------------------------------- | ------------------------------------------------------ |
| Data Loading        | `pd.read_csv()`       | Load CSV files into a DataFrame | `df = pd.read_csv("data.csv")`                         |
|                     | `pd.read_excel()`     | Load Excel files                | `df = pd.read_excel("data.xlsx")`                      |
|                     | `pd.DataFrame()`      | Create DataFrame manually       | `df = pd.DataFrame(data)`                              |
| Data Inspection     | `df.head()`           | View first rows                 | `df.head(5)`                                           |
|                     | `df.tail()`           | View last rows                  | `df.tail(5)`                                           |
|                     | `df.shape`            | Get dataset dimensions          | `df.shape`                                             |
|                     | `df.columns`          | Get column names                | `df.columns`                                           |
|                     | `df.dtypes`           | Check column data types         | `df.dtypes`                                            |
|                     | `df.info()`           | Summary of dataset              | `df.info()`                                            |
|                     | `df.describe()`       | Statistical summary             | `df.describe()`                                        |
| Column Selection    | `df["col"]`           | Select a column                 | `df["Price"]`                                          |
|                     | `df[["col1","col2"]]` | Select multiple columns         | `df[["City","Product"]]`                               |
| Row Filtering       | Boolean Filtering     | Filter rows by condition        | `df[df["Price"] > 1000]`                               |
|                     | Multiple Conditions   | Apply AND/OR conditions         | `(df["Price"]>1000) & (df["Category"]=="Electronics")` |
| Data Cleaning       | `df.isnull()`         | Detect missing values           | `df.isnull().sum()`                                    |
|                     | `df.dropna()`         | Remove missing rows             | `df.dropna()`                                          |
|                     | `df.fillna()`         | Replace missing values          | `df.fillna(0)`                                         |
| Sorting             | `df.sort_values()`    | Sort data                       | `df.sort_values("Price")`                              |
| Unique Values       | `df.unique()`         | Unique values in column         | `df["City"].unique()`                                  |
|                     | `df.nunique()`        | Count unique values             | `df["City"].nunique()`                                 |
| Frequency           | `df.value_counts()`   | Count category frequency        | `df["Product"].value_counts()`                         |
| Feature Engineering | New Column            | Create derived column           | `df["Total"]=df["Price"]*df["Qty"]`                    |
| Aggregation         | `groupby()`           | Group data for analysis         | `df.groupby("City")`                                   |
|                     | `agg()`               | Multiple aggregations           | `agg(["sum","mean"])`                                  |
| Pivoting            | `pivot_table()`       | Create pivot tables             | `pd.pivot_table(df, index="City")`                     |
| Date Handling       | `pd.to_datetime()`    | Convert column to datetime      | `pd.to_datetime(df["Date"])`                           |
| Export              | `df.to_csv()`         | Save DataFrame to CSV           | `df.to_csv("output.csv")`                              |

# One-Line Mental Model
> **Pandas workflow = Load → Inspect → Clean → Transform → Analyze → Export**

Which maps to:

| Step      | Pandas Tools                     |
| --------- | -------------------------------- |
| Load      | `read_csv()`                     |
| Inspect   | `head()`, `info()`, `describe()` |
| Clean     | `dropna()`, `fillna()`           |
| Transform | new columns                      |
| Analyze   | `groupby()`, `pivot_table()`     |
| Export    | `to_csv()`                       |
